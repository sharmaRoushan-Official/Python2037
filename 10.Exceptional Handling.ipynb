### 8. Single Except Handling Multiple Exceptions



```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    print(x/y)
except (ZeroDivisionError, ValueError):
    print("ZeroDivisionError Occurred or ValueError")
```

### Using Exception as Object (as e)

### try-except-else Block

else block runs only if no exception occurs.

    

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    value = x/y
    
except Exception as e:
    print("Error Occurred:",e)
else:
    print("Result is:",value)
```

Flow:

if exception ➽ except runs

if no exceptoin ➽ else runs

### try-except-else-finally Block

finally block always runs(wheter exception is happens or not.)

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    value = x/y
    
except Exception as e:
    print("Error Occurred:",e)
else:
    print("Result is:",value)

finally:
    print("Code Executed Successfully")
```

## 16. Custom Exceptions (User-Defined Exception)

You can Create your own excpetion class

```python
class InvalidAgeError(Exception):  # user-defined Error
    pass
```

```python
age = int(input("Enter age: "))

if age<18: # 21<18
    raise InvalidAgeError("Age must be 18 or above")  # In-Built Error
else:
    print("Age is Valid")
```

```python
## 15. Create a funciton that calculates the square root of a number. Handle ValueError if input <0.
def squareNum():
    num = int(input("Enter any number greater than 0: "))
    try:
        sqr = math.sqrt(num)
    except Exception as e:
        print("Error Occurred:",e)
    else:
        return f"Square root of value is:{sqr}"
    finally:
        print(f"Code Executed Successfully")

squareNum()
```

```python
### 15. create a menu-driven calculator using exception handling.

print("===== Calculator =====")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = input("Enter your choice: ")

try:
    num1 = float(input("Enter First Number: "))
    num2 = float(input("Enter Second Number: "))

    if choice == "1":
        result = num1 + num2
    elif choice == "2":
        result = num1 - num2
    elif choice == "3":
        result = num1 * num2
    elif choice == "4":
        result = num1 / num2
    else:
        print("Invalid Choice")

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

except ValueError:
    print("Error: Please enter valid numbers.")

else:
    if choice in ["1", "2", "3", "4"]:
        print("Result =", result)

finally:
    print("Operation Completed.")
```
