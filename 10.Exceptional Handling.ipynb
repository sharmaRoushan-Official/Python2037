### 1. What is an Exception?

An exception is an `unexcepcted event/error` that occurs during the exectution of a program.



```python
a = 10
b = 0
print(a/b) # Error
print("Hello world") # no output
```

### 3. Why Exception Handling is Required?

#### without Exception handling

* Progaram Crashes
* User gets error message
* system becomes unreliable(Not Trusted).

#### With exception handling:

* Program Continue after getting Error.
