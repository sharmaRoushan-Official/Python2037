### Q6.Count the total number of characters in a file.


```python
fileName=r"C:\Users\punam\OneDrive\Desktop\NursaryRhytham.txt"

with open(fileName, 'r')as file:
    countChar = 0
    space = 0
    for data in file.read():
        if data == " ":
            space += 1
            countChar += 1
    print("total number of character in file:",countChar)
    print("total number of space is:",space)
    print("total number of character without space is:",countChar-space)
```

### 12. Count the number of uppercase and lowercase letters in a file.


```python
fileName = r"C:\Users\punam\OneDrive\Desktop\NursaryRhytham.txt"

upperChar = 0
upperCharacter = []

lowerChar = 0
lowerCharacter = []

with open(fileName,'r') as file:
    for ch in file.read():
        if ch.isupper():
            upperCharacter.append(ch)
            upperChar+=1
        else:
            if ch == " "or ch =="\n"or ch == "\t":
                continue
            lowerChar +=1
            lowerCharacter.append(ch)

print("upper character is:",upperCharacter,"=",upperChar)
print("lower character is:",lowerCharacter,"=",lowerChar)
```
