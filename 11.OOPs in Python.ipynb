## 11. OOPs in Python [Object-Oriented programming structure]

### 1.What is OOP?

  * OOP is a programming paradigm [A programming paradigm is a fundamental style or "way of thinking" about how you structure and write your code.] that organizes software design around objects, rather than functions and logic.


### Why OOP?

 * Code reusability
 * Modularity  [Modularity in Object-Oriented Programming (OOP) is the practice of breaking a complex software program into small, independent, and self-contained units of functionality]
 * Scalability [Scalability in OOP refers to a system's ability to seamlessly handle growth—whether it's increased traffic, data volume, or new features—without crumbling]
 * Easy Maintainace
 * Real-world modeling

### 2. Class

A `class` is a blueprint or template used to create any object.

syntax:

````python
class Student:
    pass
````

## Key Points:

* Defined using `class` keyword
* Can contains variables and method.
* Naming Convention: PascalCase

````python
class Student:  # Student is the name of class 
    def display(self):  # display is the method name
        print("Welcome to Python OOP")  # statement
````

````python
s1 = Student()
s1.display()
````

````python
s2 = Student()
s2.display()
````

### 3. Object 

An `Object` is an instance of a class.

syntax:

````python
s1 = Student()
````

## Keypoint

* Memory allocated when object is created.
* Can access class method and variables.

## 4. self Keyword

####  Define:

self refers to the currect object of class.

### ➽ Why Required?

* To access instance variables and methods.

````python
class Demo:
    def show(self):
        print("Self represents currect object")
````

### 5. Constructor (`__init__`)

A `constructor` initializes  object data automatically when the object is created.

# Instance, Attribute, and Method in Python

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance Method
    def studentDetails(self):
        print("Student Name:", self.name)
        print("Student Age:", self.age)

obj = Student("Avnish", 13)

```

# 2. Instance (Object)

An **instance** (also called an **object**) is a real object created from a class.

```python
obj = Student("Avnish", 13)
```

Here,

- `obj` is the **instance (object)** of the `Student` class.

### Structure

```text
Student (Class)
      │
      ▼
obj (Instance/Object)
```

---

# 3. Attributes (Instance Variables)

Attributes are variables that belong to an object.

```python
self.name = name
self.age = age
```

Here,

- `self.name` → Attribute (Instance Variable)
- `self.age` → Attribute (Instance Variable)

After creating the object,

```python
obj = Student("Avnish", 13)
```

the object stores:

```text
obj
├── name = "Avnish"
└── age = 13
```

You can access them as:

```python
obj.name
obj.age
```

Output:

```text
'Avnish'
13
```

---

# 4. Instance Method

An **instance method** is a method that works with an object and has `self` as its first parameter.

```python
def studentDetails(self):
```

It accesses the object's attributes using `self`.

Called as:

```python
obj.studentDetails()
```

Output:

```text
Student Name: Avnish
Student Age: 13
```

---

# 5. Constructor

A **constructor** is a special method that is automatically called when an object is created.

```python
def __init__(self, name, age):
```

Purpose:

- Initializes the object.
- Assigns values to the instance variables (attributes).

---

# 6. self Keyword

`self` refers to the **current object (instance)**.

When Python executes

```python
obj.studentDetails()
```

Python automatically converts it to

```python
Student.studentDetails(obj)
```

So,

- `self` = `obj`

### 6. Types of Variables

#### 1. Instance Variables
- Defined using `self`
- Object-specific (each object has its own copy)

**Example:**
```python
class Student:
    def __init__(self):
        self.name = "John"
```

---

#### 2. Class Variables
- Shared among all objects
- Defined directly inside the class but outside any method

**Example:**
```python
class College:
    name = "ABC College"
```

   

### 1. Instance Variables

````python
class Student:  # Student is class anme
    def __init__(self,name,age): # Constructor
        self.name = name  # Attribute of object
        self.age = age  # Attribute object

    def studentDetails(self):  # method of class
        print(f"Student name is:{self.name}\nStudent Age is:{self.age}")
obj = Student("Avnish",13)
obj.studentDetails()
````

### 2. Class Variables

````python
class Student:  # Student is class anme
    schoolName = "SSD College"  # class variable
    def __init__(self,name,age): # Constructor
        self.name = name  # Attribute of object
        self.age = age  # Attribute object

    def studentDetails(self):  # method of class
        print(f"Student name is:{self.name}\nStudent Age is:{self.age}")
obj = Student("Avnish",13)
obj.studentDetails()
````
