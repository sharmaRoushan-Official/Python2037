## 11. OOPs in Python [Object-Oriented programming structure]

### 1.What is OOP?

  * OOP is a programming paradigm [A programming paradigm is a fundamental style or "way of thinking" about how you structure and write your code.] that organizes software design around objects, rather than functions and logic.


### Why OOP?

 * Code reusability
 * Modularity  [Modularity in Object-Oriented Programming (OOP) is the practice of breaking a complex software program into small, independent, and self-contained units of functionality]
 * Scalability [Scalability in OOP refers to a system's ability to seamlessly handle growth—whether it's increased traffic, data volume, or new features—without crumbling]
 * Easy Maintainace
 * Real-world modeling

### 2. Class

A `class` is a blueprint or template used to create any object.

syntax:

```python
class Student:
    pass
```

## Key Points:

* Defined using `class` keyword
* Can contains variables and method.
* Naming Convention: PascalCase

```python
class Student:  # Student is the name of class 
    def display(self):  # display is the method name
        print("Welcome to Python OOP")  # statement
```

```python
s1 = Student()
s1.display()
```

```python
s2 = Student()
s2.display()
```

### 3. Object 

An `Object` is an instance of a class.

syntax:

```python
s1 = Student()
```

## Keypoint

* Memory allocated when object is created.
* Can access class method and variables.
