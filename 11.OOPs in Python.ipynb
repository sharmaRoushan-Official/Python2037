## a). Single/Simple Inheritance

##### Meaning
 * In `Single Inheritance`, a child inherits from only one parent class.

📌 Structure

`parent ➼ Child`

### Why It is used:
 * Code resability
 * Child class can use parent class properties and method.
 * Easy to understand and maintain.

### Real-life Example

* Parent- Vehicle
* Child - Car

A Car is a Vehicle


Python Example

````python
class Vehicle:  # parent
    def start(self):
        print("Vehicle Started.")

class Car(Vehicle):   # child of vehicle
    def drive(self):
        print("Car is driving.")


# obj = Vehicle()
# obj.start()

# obj1 = Car()
# obj1.drive()


obj2 = Car()
obj2.start()  # start is a method of parent calling from child
obj2.drive()  # drive is the method of child calling from child.
````

Note:

Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.

[ambiguity occurs when the compiler or runtime environment encounters two or more valid ways to interpret a method call, variable, or object reference, leaving it unable to decide which one to execute]

## 2. Multiple Inheritance

Meaning:

In `Multiple Inheritance`, a child class inherits from more than one parent class.

📌 Structure

```text
   Parent1          Parent2
       \            /
        \          /
         \        /
           Child
```

### Why It is used?

* When one class needs feature from multiple classes
* Helps to combin functionalities

### Real-life Example
 * Parent1: Camera
 * Parent2: Phone
 * Child: SmartPhone

SmartPhone is both a Phone and a Camera.


Python Example:



````python
class Phone:  # parent
    def call(self):
        print("Calling.....")

class Camera:  # parent
    def click_photo(self):
        print("Photo Clicked....")

## multiple Inheritance
class SmartPhone(Phone,Camera): # SmartPhone is the child of Phone and Camera
    def internet(self):
        print("Google Connected.")

obj = SmartPhone()
obj.call()
obj.click_photo()
obj.internet()
````

Note:
 Python support multiple inheritance, You can check order of method called Method Resolution Order (MRO) to avoid confusion

````python
print(SmartPhone.mro())
````

### 3. Mutilevel Inheritance

Meaning

In `Multilevel Inheritance`, inheritance happens in levels.

📌 Structure

`GrandParent ➼ Parent ➼ Child`


### Why It is used?
* To represent hierarchy in real world.
* Step-by-step inheritance of features

## Real-life Example

* Grandparent: Animal
* Parent: Mammal
* Child: Dog

Dog is a Mammal, Mammal is an Animal.


Python Example:

````python
class Animal:  # parent
    def eat(self):
        print("Eating...")

class Mammal(Animal):  # Child of Animal
    def walk(self):
        print("Walking....")

class Dog(Mammal):  # Child of Mammal
    def bark(self):
        print("Barking.....")

d = Dog()
d.eat()
d.walk()
d.bark()
````

Note: Child class gets access to all previous level methods.

````python
print(Dog.mro())
````

### 4. Hierarchical Inheritance

Meaning:

In `Hierarchical Inheritance`, Multiple child classes inherits from a single parent class.


📌 Structure

```text
                 Parent1
              /     |     \
         child1   child2   child3
```


## Why it is used?

* One parent has common functionality
* Different children use and extend it differently.


### Real-life Example

* Parent: Employee
* Children:
    * Manager
    * Developer
    * Tester
* All are employees but have different roles.

### Python Example


````python
class Employee:
    def work(self):
        print("Employee is working.....")


class Manager(Employee):
    def manage(self):
        print("Manager is managing work.....")

class Developer(Employee):
    def code(self):
        print("Developer is coding.......")

class Tester(Employee):
    def test(self):
        print("Tester is Testing......")


m = Manager()
m.work()  # parent class method 
m.manage()  # manager metthod

d = Developer()
d.work() # parent class method
d.code() # developer class method


t = Tester()
t.work()  # parent class method
t.test() # Terster class method 
````

````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

### 5. Hybrid Inheritance:

Meaning:

Hybrid Inheritance is combination of two or more inheritance types.

It May Include:
 * Multiple + Multilevel
 * Hierarchical + Mutiple

📌 Structure

```text
              A
            /   \
           B     C
            \   /
              D
```
Here:
* B and C inherits from A (Heirarchical Inheritance)
* D inherits from B and C (Multiple Inheritance)

### Why it is used?
 * When a system requires complex relationships.
 * Common in large framework and enterprises projects.

# Python Example

````python
class A:
    def showA(self):
        print("Class A")

class B(A):  # B is the child of A
    def showB(self):
        print("Class B")

class C(A):  # C is the child of A
    def showC(self):
        print("Class C")
class D(B,C):  # D is the child of B and C
    def showD(self):
        print("Class D")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()

# obj.
````

Note:

Hybrid Inheritance is powerful but can cuase confusing due to 

* Method conflicts
* ambiguity

That's why python uses MRO (Method Resoluction Order):

````python
print(D.mro())
````

|Access Modifier|Synatx|Meaning|

|Public|`variable`|Accessible everywhere|
|Protected|`_variable`|Accessible in class & child|

### 1. Public Members


````python
class Employee:
    def __init__(self,name):
        self.name = name ## Public varaible

emp = Employee("Ram")
print(emp.name)  # varible is accessible outside the class 
````

### 2. Protected Members (`_`)

````python
class Employee:
    def __init__(self,salary):
        self._salary = salary # Protected Variable


class Manager(Employee):
    def show_salary(self):
        print(self._salary)  # Protected Varaible

m = Manager(50000)
m.show_salary()
````

````python
e = Employee(10000)
e._salary  # protected
````

Notes:

* Convention based
* Should not be accessed directly outside class

### 3. Private Members (`__`)

````python
class BankAccount:
    def __init__(self):
        self.__balance = 1000  # Private Variable

    def show_balance(self):
        print("Balance:",self.__balance)

    def withdraw(self,amount):
        if amount<=self.__balance:
            self.__balance -= amount
            print("Withdrawal Successful")
        else:
            print("Insufficient balance")
````

````python
acc = BankAccount()
acc.show_balance()
````

````python
acc.withdraw(200)
````

````python
acc.show_balance()
````

````python
acc.__balance
````

### How Private Works internally

Python uses Name Mangling [Name mangling is a compile-time mechanism where Python automatically renames any class identifier starting with a double underscore (e.g., __variable) to _ClassName__variable to prevent accidental naming collisions during inheritance.]

````python
print(acc._BankAccount__balance)
````

### Encapsulation using Getter & Setter methods

````python
class Student:
    def __init__(self):
        self.__marks = 0

    ## setter
    def set_marks(self,marks):
        if marks>=0 and marks<=100:
            self.__marks = marks
        else:
            print("Invalid Marks")
    # getter: display value on the screen
    def get_marks(self):
        return self.__marks
````

````python
s = Student()
s.set_marks(85)  # 85 in marks [setter]
````

````python
s.get_marks()  # getter method
````

````python
s = Student()
s.set_marks(101)  # 85 in marks [setter]  # data validation
````

## Notes:

* Data Validation
* Controlled access

### Encapsultaion using @property Decorator

````python
class Employee:
    def __init__(self):
        self.__salary = 0
        
    @property
    def salary(self):
        return self.__salary
        
    @salary.setter
    def salary(self,amount):
        if amount>0:
            self.__salary = amount
        else:
            print("Invalid Salary")
````

````python
emp = Employee()
emp.salary = 4000  # setter
````

````python
emp.salary  # property [getter]
````

### 4. Data Abstraction (OOPS Concept)

##### Meaning of Data Abstraction

Data Abstraction means hiding internal implementataion details and showing only essential features to the user.

📌 Simple definition 

> What an object does is shown, how it does is hidden.

### Why Data Abstraction is Needed 

#### Why Data Abstraction is Needed

❌ Without Abstraction

* User sees unnecessary details
* Code becomes complex
* Diffcult to maintain
* Higher chances of misuse
  
✅ With Abstraction:
* Only required features exposed
* Internal logic hidden
* Easy to use & maintain
* Improves security

### Real-Life Example

##### Driving a Car
* You can use:
  * Sterring
  * Accelerator
  * Break

###### You don't know engine internals
> This is Data Abstraction

#### Data Abstraction VS Encapsulation

----
|Feature|Encapsulation|Abstraction|
|----|------|----|
|`Focus`|Data hiding|Implementation hinding|
|`Shows`|How data is protected|What functionalrity is provided|
|`Achieved by`|Access modifiers|Abstract classes/interfaces|

📌 Structure

> Encapsulation = How data is stored

> Abstraction = What user can do


### How Abstraction is Achieved in Python

Python Provides: 1. Abstract Base Classes(ABC) 2.Abstract Methods

### 1. Abstract Base Class(Core Concept)

📌 An abstract base class:
* Cannot be instantiated.
* Contains abstract methods [Built-in]
* Acts as a blueprint


### 2. Abstract Method
📌 An abstract method:
* Has declaration only
* No implementation
* Must be implemented by child class

### Syntax for Abstraction in Python

````python
from abc import ABC, abstractmethod
````

### Basic Abstraction Example



````python
class Vehicle(ABC):

    @abstractmethod    ## permanently hidden
    def start(self):  # start method 
        pass
    @abstractmethod
    def abcd(self):
        pass
# v = Vehicle()
# c.start()  # unable to display method throw error: can't created as object
````

````python
class Car(Vehicle):
    def automaticCar(self):
        print("My Car is automatic.")

    def start(self):
        print("Car is Started")

    def abcd(self):
        pass

c = Car()
c.start()  # parent method
c.automaticCar()  # child method
````

### Key-Points

* Vechicle defines what should be done.
* Car & Bike define how it is done.
* User only calls start()
### Rules of Data Abstraction

* ✓ Abstract class must be inherit ABC3
* ✓ Abstract mehtod must use @abstractmethod
* ✓ Child class must implement abstract methods
* ✓ Abstract class object cann't be created.

### When to Use Abstraction

* Use abstraction when:
    * Common Functionality exists
    * Multiple implementations needed
    * System design is complex
    * Framework or lagre application

#### One-Line [Interview]
* ✓ Abstraction hides implementation details
* ✓ Achieved using abstract classes

### Super() constrcutor

### What is super()?

super() is a built-in python function used to call method or the constrcutor of the parent(base) class from a child(dervied) class.

* It is mainly used with inehritance.

#### Syntax

`super().__init__(arguments)`


### Why use super()?

* Calls the parent class constructor.
* Resuses parent class code.
* Avoids rewriting code.
* Supports multiple inheritance.
* Makes code easier to maintain.


### Advantages

* Code Reusability
* Cleaner code
* Easier Maintainance
* Supports Multiple Inheritance
* Avoids Duplicate Code.

### Parent Constructor without `super()`

````python
class Person:
    def __init__(self):  # Constructor
        print("Parent Constructor")
class Employee(Person):
    def __init__(self):  # Constructor
        print("Employee Constructor")

obj = Employee()
obj
````

Note:
Parent constructor is not called.

### Parent Constructor Using `super()`

````python
class Person:
    def __init__(self):
        print('Person Constructor')

class Employee(Person):  # Employee Class
    def __init__(self):  # Employee Constructor
        super().__init__()  # Parent(Person) Constructor
        print("Employee constructor") # Employee Statement
obj = Employee()
obj
````

### Passing Argumetns 

````python
class Student:
    def __init__(self,name):
        self.name = name
obj = Student("Ravi")
obj.name
````

````python
class Student2(Student):
    def __init__(self,name,age):
        # self.name = name
        self.age = age
        super().__init__(name)

obj1 = Student2("Ravi",32)
print(obj1.name)
print(obj1.age)
````

````python
class Student3(Student2):
    def __init__(self,name,age,email):
        # self.name = name
        # self.age = age
        self.email = email
        super().__init__(name,age)
obj3 = Student3("Ravi",32,"ravi@gmail.com")
print(obj3.name)
print(obj3.age)
print(obj3.email)
````

### Calling Parent Method



````python
class Person:  # parent class
    def show(self):  # show is method
        print("I am person")  # method statement

class Employee(Person):  # Employee is child
    def show(self):  # show is the method of Employee
        super().show() # here show() is the method of parent
        print("I am an Employee") # Employee statement

obj = Employee()
obj.show()
````

# Constructor

### Definition
A **constructor** is a special method in Python that is **automatically called whenever an object (instance) of a class is created**. It is used to initialize the object's data (instance variables).

### Syntax
```python
class ClassName:
    def __init__(self):
        # Initialization code
```

### Example
```python
class Student:
    def __init__(self, name, age):
        print("Constructor Called")
        self.name = name
        self.age = age

# Object Creation
s1 = Student("Ravi", 25)
```

### Output
```
Constructor Called
```

### How It Works
1. A class is created.
2. An object is created.
3. Python automatically calls the `__init__()` constructor.
4. The constructor initializes the object's instance variables.

### Key Points
- `__init__()` is the constructor in Python.
- It is automatically called when an object is created.
- It is used to initialize object data.
- It is executed once for each object.
- You do not need to call it manually.

### Remember
**Class → Object Created → Constructor (`__init__`) Called Automatically → Object Initialized**

### Interview Definition
> **A constructor is a special method (`__init__`) that is automatically called when an object of a class is created. It is used to initialize the object's data.**

## Q.9 Create a class Employee and override its `__str__()` method to print custom details when the object is printed.




````python
class Employee:
    def __init__(self,emp_id,name,department,salary):
        self.emp_id = emp_id
        self.name = name
        self.department = department
        self.salary = salary

    def __str__(self): # __str__() is a magic (dunder)
        return (f"Employee ID:{self.emp_id}\n"
               f"Name:{self.name}\n"
        f"Department:{self.department}\n"
        f"Salary:{self.salary}")
# creating Object 

emp1 = Employee(101,"Singh Shaab Great","IT",5000)
print(emp1)
````

# Python Dunder (Magic) Methods

## What are Dunder Methods?

- **Dunder** stands for **Double UNDERscore**.
- Dunder methods are **special methods** in Python whose names start and end with **double underscores (`__`)**.
- They allow you to define how objects behave with built-in operations.
- Most dunder methods are **automatically called by Python**.

**Syntax**
```python
def __method_name__(self):
    pass
```
