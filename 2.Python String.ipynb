### 2. Python String


### 1. What is a string in Python?

A `string` in Python is sequence of Unicode character enclosed in quotes.

```python
s1 = "Hello"
s2 = "World"
s3 = """Multi-line String"""
print(s1)
print(s2)
print(s3)
```

```python
ord("H")
```

#### Key Points:

* Strings are immutable(cannot be changed after creation)
* Indexed(Ordered collections)
* Supports Slicing, iteration, and many built-in methods

## 2. Creating Strings

##### Different Way


```python
s1 = "Hello"  # double quotes
s2 = 'Hello'  # single quotes
s3 = """Hello world"""  # triple quotes

print(s1)
print(s2)
print(s3)
```

### 4. Slicing of String
----

```python
x = "Hello world"
print(x)

x1 = x[0:5]
print(x1)

x2 = x[6:11]
print(x2)
```

```python
x = 'Hello world'
print(x)


x1 = x[0:5]
print(x1)

x2 = x[6:11]
print(x2)
```

```python
x = 'Hello world'
print(x)

x3 = x[:5]
print(x3)

x4 = x[6:]
print(x4)
```

### Striding of String 

----

syntax:

string[starts:stops:skip]

```python
x = "Hello world"
print(x)

x1 = x[0:11:2]  # 1 characters skip
print(x1)
x2 = x[0:11:3]  # 2 characters skip
print(x2)
```

```python
# Reverse values

x = "Hello world"
print(x)

x1 = x[::-1]  # Reverse printing
print(x1)
```

### Important String Methods

------

🔷 Case Conversion

---

```python
s = "hEllO wORld"  # random case
print("Original Value:",s)

# upper()

value = s.upper()
print(value)

# lower()
value2 = s.lower()
print(value2)

# capitalize()
value3 = s.capitalize()
print(value3)

# title()

value4 = s.title()
print(value4)

# swapcase() : only change the case of original value

value5 = s.swapcase()
print(value5)
```

### 🔷 Searching & Finding

```python
s = "Hello World"
print(s)

print(s.find("a"))  # if not character found in the string it will return -1
print(s.index('a'))  # if no character found in the string it will return error
```

```python
s = "Hello World"
print(s)

print(s.find("H"))  # if characters found in the string it will return 0
print(s.index("H")) # if characters found in the string it also will return 0
```
