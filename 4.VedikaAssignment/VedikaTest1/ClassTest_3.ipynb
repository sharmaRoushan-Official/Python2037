### 4️⃣ For Loop
### 4 .Write a program to print the multiplication table of a given number from 1 to 10.


```python
for i in range(1,11):
    print(f"----Table of{i}----")
    for j in range(1,11):
        print(i,"x",j,"=",i*j)
    print()        
```

### 5️⃣ For Loop
### 5.  Write a program to find the sum of all even numbers between 1 and N.


```python
n = int(input("Enter N:"))
sum = 0
for i in range(2,n+1,2):
    sum = sum + 1
print("Sum of evn numbers=", sum)    
```

### 6️⃣ For Loop
### Write a program to find the factorial of a number using a "for" loop.


```python
n = 20
fact = 1
for i in range(1, n+1):
    fact = fact*i
print("Factorial=", fact, end=" ")    
```

### 7️⃣ While Loop
### 7. Write a program to reverse a given number using a "while" loop.
Example: "12345 → 54321"


```python
num = int(input("Enter number:"))
rev = 0

while num > 0:
    digit = num % 10
    rev = rev*10 + digit
    num = num // 10
print("Reverse =",rev)
```

### 8️⃣ While Loop
### 8. Write a program to count the number of digits in a given number using a "while" loop.


```python
num = int(input("Enter the number:"))
count = 0
while num > 0:
    count += 1
    num = num // 10
print("Digits:", count)    
```

### ️9️⃣ Break
### 9. Write a program to take numbers from the user continuously and stop taking input when the user enters 0. Finally, print the sum of all entered numbers.


```python
sum = 0
while True:
    num = int(input("Enter a number:"))
    if num == 0:
        break
    sum = sum + num
print("Sum=", sum)    
```

### 11. Create a function "is_prime(n)" that accepts a number and returns whether the number is prime or not.


```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True
num = int(input("Enter a number:"))    
    
if is_prime(num):
    print("Prime number")
else:
    print("Not prime number")
```

### 1️⃣5️⃣ OOP
### 15. Create a class "Student" with:
* Attributes: "name", "marks"
* Method: "display()"
* Method: "check_result()"

#"check_result()" should print "Pass" if marks are 40 or above, otherwise "Fail".

 Create an object and test the program with user-provided values.

```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
    def check_result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")
name = input("enter student name:")
marks = int(input("Enter marks:"))
s = student(name, marks)

s.display()
s.check_result()
```
