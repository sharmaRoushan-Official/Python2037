#5. Write a program to count the number of vowels in a string.

```python
s=("vedika")
count=0
for ch in s.lower():
    if ch in "aeiou":
        count+=1
        print(count)
```

#7. Write a program to check whether a given string is a palindrome.

```python
s=("vedika")
if s==s[::-1]:
    print("palindrime")
else:
    print("not palindrme")
```
