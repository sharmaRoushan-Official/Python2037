#21. Write a Python program to calculate *discount* based on purchase amount:

    * > 1000 → 20% discount
    * 500–1000 → 10% discount
    * <500 → No discount

```python
amount =float(input())
if amount > 1000:
    discount = amount * 0.20
elif amount >= 500:
    discount =amount * 0.10
else:
    discount = 0
print("Discount =", discount)
print("Final Amount =", amount-discount)
```
