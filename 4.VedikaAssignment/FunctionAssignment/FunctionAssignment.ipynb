#### 10. Write a function that takes a list and returns it without duplicates.

```python
def duplicate(list):
    result = []
    for i in list:
        if i not in result:
            result.append(i)
    return result        
numbers = [1,2,2,3,4,4,5,5]
print(duplicate(numbers))
```

#### 16. Write a function that accepts a list of integers and returns the sum of even numbers.

```python
def even(list):
    s = 0
    for i in list:
        if i % 2 == 0:
            s = s + i
    return s
print(even([1,2,3,4,5,6]))    
```

#### 17. Write a function that takes a sentence and returns the number of words.


```python
def words(sentence):
    count = 1
    for ch in sentence:
        if ch == " ":
            count += 1
    return count
print(words("Python is programming language"))    
```

#### 18. Write a function that takes a number and returns the sum of its digits.

```python
def sum(n):
    s = 0
    while n > 0:
        digit = n % 10
        s = s + digit
        n = n // 10
    return s
print(sum(1234))    
```

####  26. Write a function that takes variable number of arguments and returns their average.

```python
def average(*args):
    if len(args) == 0:
        return 0
    total = sum(args)
    return total / len(args)
print(average(10,20,30,40))    
```

#### 27. Write a function to remove punctuation from a string.


```python
def punctuation(s):
    result = ""
    for ch in s:
        if ch.isalnum() or ch.isspace():
            result += ch
    return result        
print(punctuation("Hello, world! How are you?"))            
```

#### 29. Write a recursive function to calculate factorial.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(5))        
    
```

#### 30. Write a recursive function to compute the nth Fibonacci number.

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci (n - 1) + fibonacci(n - 2)
print(fibonacci(6))        
```

#### 31. Write a function to check if a given year is a leap year.

```python
def leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return True
    else:
        return False
print(leap_year(2024))        
        
```

#### 33. Write a function that returns all prime numbers up to a given number n.

```python
def primes_upto(n):
    primes = []
    for num in range(2, n + 1):
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            primes.append(num)
    return primes
print(primes_upto(20))
```

#### 34. Write a function that takes a list and returns a new list with only unique elements (preserving order).

```python
def unique_list(list):
    result = []
    for item in list:
        if item not in result:
            result.append(item)
    return result
print(unique_list([1,2,2,3,4,4,5]))    
```

#### 44. Write a decorator function that logs the execution time of another function.

```python
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(end - start)
    return wrapper
@timer
def hello():
    print("Hello")
    time.sleep(1)
hello()    
```

#### 45. Write a recursive function to find the sum of all elements in a list.

```python
def list_sum(lst):
    if len(lst) == 0:
        return 0
    return lst[0] + list_sum(lst[1:])
print(list_sum([1,2,3,4,5]))    
```
