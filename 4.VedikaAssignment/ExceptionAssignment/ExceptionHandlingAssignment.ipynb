### *15. Try-Except Inside a Function*

Create a function that calculates the square root of a number. Handle ValueError if input < 0.

```python
import math
```

```python
def squareNum():
    num = int(input("Enter any number greater than 0:"))
    try:
        sqr = math.sqrt(num)
    except Exception as e:
        print("Error occurred:",e)
    else:
        return f"Square root of value is:{sqr}"
    finally:
        print(f"Code Excuted Successfully")
squareNum()        
```

#### *16. Reading From a Possibly Corrupt File*

Write a program that tries to read and convert each line of a file into an integer. Handle:

* FileNotFoundError
* ValueError (line cannot convert)

```python
try:
    with open("numbers.txt", "r") as file:
        for line in file:
            try:
                num = int(line.strip())
                print(num)
            except ValueError:
                print("Cannot convert:", line.strip())
except FileNotFoundError:
    print("File not found.")
```
