```python
# for loop

numbers = [10,15,22,33,40,51,60]
even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("Even numbers:",even)
print("Odd number:",odd)
```

#### 2. Find the second largest number in a list without using sort() or max().

```python
# for loop
lst = [10, 25, 40, 15, 30]

first = lst[0]
second = lst[0]

for i in lst:
    if i > first:
        second = first
        first = i
    elif i > second and i != first:
        second = i

print("Second Largest:", second)
```

```python
# while loop
lst = [10, 25, 40, 15, 30]

first = lst[0]
second = lst[0]

i = 0
while i < len(lst):
    if lst[i] > first:
        second = first
        first = lst[i]
    elif lst[i] > second and lst[i] != first:
        second = lst[i]
    i += 1

print("Second Largest:", second)
```

#### 16. Print a number diamond pattern.

```python
# for loop
n = int(input("Enter number of rows: "))

# Upper half
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end=" ")
    for k in range(2 * i - 1):
        print(i, end=" ")
    print()

# Lower half
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end=" ")
    for k in range(2 * i - 1):
        print(i, end=" ")
    print()
```

```python
# while loop
n = int(input("Enter number of rows: "))

i = 1
while i <= n:
    j = 1
    while j <= n - i:
        print(" ", end=" ")

        j += 1

    k = 1
    while k <= 2 * i - 1:
        print(i, end=" ")
        k += 1

    print()
    i += 1

i = n - 1
while i >= 1:
    j = 1
    while j <= n - i:
        print(" ", end=" ")
        j += 1

    k = 1
    while k <= 2 * i - 1:
        print(i, end=" ")
        k += 1

    print()
    i -= 1
```

#### 21. Print Pascal's Triangle.

```python
# for loop
n = int(input("Enter number of rows: "))

for i in range(n):
    num = 1

    for s in range(n - i - 1):
        print(" ", end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()
```

```python
# while loop
n = int(input("Enter number of rows: "))

i = 0
while i < n:
    num = 1

    s = 0
    while s < n - i - 1:
        print(" ", end="")
        s += 1

    j = 0
    while j <= i:
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
        j += 1

    print()
    i += 1
```

#### 25. Implement Selection Sort using for loops.

```python
# for loop
list = [2,5,4,3,1]
for i in range(len(list)):
    for j in range(i + 1,len(list)):
        if list[i] > list[j]:
            list[i],list[j] = list[j],list[i]  
print(list)                                    
```

```python
# while loop
list = [2,5,4,3,1]
i = 0
while i < len(list):
    j = i + 1
    while j < len(list):
        if list[i] > list[j]:
            list[i],list[j] = list[j],list[i]
        j += 1
    i += 1
print(list)    
      
```

#### 26. Implement Bubble Sort using nested for loops.

```python
# for loop
list = [5, 2, 8, 1, 3]

n = len(list)

for i in range(n):
    for j in range(n - 1 - i):
        if list[j] > list[j + 1]:
            temp = list[j]
            list[j] = list[j + 1]
            list[j + 1] = temp

print("Sorted List:", list)
```

```python
# while loop

list = [5, 2, 8, 1, 3]

n = len(list)

i = 0
while i < n:
    j = 0
    while j < n - 1 - i:
        if list[j] > list[j + 1]:
            temp = list[j]
            list[j] = list[j + 1]
            list[j+1] = temp
        j += 1
    i += 1

print("Sorted List:", list)
```
