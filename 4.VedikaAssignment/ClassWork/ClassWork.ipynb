#### 11. Count the number of vowels present in a text file.


```python
fileName = r"C:\Users\VEDIKA\Desktop\NursaryRhymes.txt"
count = 0
vowels = "aeiou"
with open(fileName,'r') as file:
    for line in file:
        for ch in line:
            if ch in vowels:
                count += 1
print(count) 
```

#### 12. Count the number of uppercase and lowercase letters in a file.


```python
fileName = r"C:\Users\VEDIKA\Desktop\NursaryRhymes.txt"
upperChar = 0
upperCharacter = []
lowerChar = 0
lowerCharacter = []

with open(fileName, 'r') as file:
    for ch in file.read():
        if ch.isupper():
            upperCharacter.append(ch)
            upperChar += 1
        else:
            if ch == " " or ch == "\n" or ch == "\t":
                continue
            if ch.islower():
                lowerChar += 1
                lowerCharacter.append(ch)
print("Total character is:", upperCharacter,"=", upperChar)
print("Total character is:", lowerCharacter, "=", lowerChar)
```
