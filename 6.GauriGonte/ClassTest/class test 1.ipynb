### 1 Write a program to input a number and check whether it is positive, negative, or zero.


```python
num = int(input("Enter a number:"))
if num > 0:
    print("Positive")
else:
    if num < 0:
        print("negative")
    else:
        print("zero")
```

### 2 Condition
Write a program to input three numbers and find the largest number using "if-elif-else".


```python
num1 = int(input("Enter a 1st number:"))
num2 = int(input("Enter a 2nd number:"))
num3 = int(input("Enter a 3rd number:"))

if num1 >= num2 and num1 >= num3:
    print("largest number is:", num1)
elif num2 >= num1 and num2 >= num3:
    print("largest number is:", num2)
else:
    print("largest number is:", num3)
```

### 3 Condition
Write a program to input a year and check whether it is a leap year or not.


```python
year = int(input("Enter a year:"))
if year % 4 == 0:
    print("Leap year")
else:
    print("not leap year")
```

### 5 For Loop
Write a program to find the sum of all even numbers between 1 and N.


```python
n=int(input("Enter a number:"))
a=0
for i in range(1,n+1):
    if i % 2 ==0:
        a=a+i
        print("sum=",a)
```

### 6 For Loop
Write a program to find the factorial of a number using a "for" loop.


```python
num=int(input("Enter a number:"))
fact=1
for i in range(1,num+1):
    fact=fact*i
    print("factorial=",fact)
```

### 7 While Loop
Write a program to reverse a given number using a "while" loop.
Example: "12345 → 54321"



```python
num=int(input("Enter a number:"))
rev=0
while num>0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
    print("reverse number=",rev)
```

### 8  While Loop
Write a program to count the number of digits in a given number using a "while" loop.


```python
num=int(input("Enter a number:"))
count=0
while num>0:
    count=count+1
    num=num//10
    print("number of digits=",count)
```

### 9  Break
Write a program to take numbers from the user continuously and stop taking input when the user enters 0. Finally, print the sum of all entered numbers.


```python
sum = 0

while True:
    n = int(input("Enter a number: "))
    if n == 0:
        break
    sum = sum + n

print("Sum =", sum)
```

### 10  Continue
Write a program to print numbers from 1 to 50, but skip all numbers that are divisible by 5 using "continue".


```python
for i in range(1,51):
    if i % 5 == 0:
        continue
    print(i)
```

### 14  Exception Handling
Write a program to input an integer and print its square. If the user enters a non-numeric value, handle the exception and display an appropriate message.
    

```python
try:
    num = int(input("Enter an integer:"))
    square = num*num
    print("Square =", square)

except ValueError:
    print("Invalid input, please enter a number!")
```

### 15 OOP
Create a class "Student" with:
* Attributes: "name", "marks"
* Method: "display()"
* Method: "check_result()"

"check_result()" should print "Pass" if marks are 40 or above, otherwise "Fail".

Create an object and test the program with user-provided values.

```python
class student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def check_result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")


name = input("Enter student name:")
marks = int(input("Enter marks:"))

s = student(name, marks)
s.display()
s.check_result()
```
