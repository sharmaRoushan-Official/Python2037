## 5. Sets in Python

---
1. Introduction to Sets

A `set` in Python is a collection data type used to store multiple unique elements.

It is based on the mathematical concept of a set

### Key Characteristics

1. Unordered
2. Unindexed
3. Mutlbale(can add/remove items)
4. Does not allow duplicates value
5. Elements must be immutable(hashable)

```python
# how to create empty set

s = {}
print(s)
print(type(s))

# by default {} brakcet is dictionary data type. if you create empty set then use set() method
```

```python
s2 = set()
print(s2)
print(type(s2))
```

```python
# if you put element in {} with at least one value then it is set.


s = {1}
print(s)
print(type(s))
```

### 2. Creating Sets

---
Method 1: Using Curly braces {}

```python
s = {1,2,3,4,5,6}
print(s)
print(type(s))
```

Method 2: Using set() constructor


```python
s = set([1,1,1,1,2,2,2,2,3,6,6,96,9,9,9,9,])
print(s)
print(type(s))
```

### Important

Empty set cannot be created using {}

```python
s = {}
print(s)
print(type(s))
```

correct way

```python
s = set()
print(s)
print(type(s))
```

### 3. Duplicate Elements

Sets automatically remove duplicates.

    

```python
s = {10,20,30,40,50,10,40,40,40,40}
print(s)
```

### 4. Data Types allowed in Sets

Elements must be immutable(hasable)

Allowed

* int
* float
* string
* tuple
* boolean
* frozenset

```python
s = {10,20.5,"Python",True,False,("Hello world",-100)}
print(s)
```

```python
s = {10,20.5,"Python",True,False,("Hello world",-100),[20]}
print(s)
```

### Not Allowed

Mutable types:

* list,
* dictionary,
* set

```python
s = {10,20.5,"Python",True,False,("Hello world",-100),{'a':'apple'}}
print(s)
```

```python
s = {10,20.5,"Python",True,False,("Hello world",-100),{10,20}}
print(s)
```

### 6. Checking Membership

Use in operator or not in

```python
s = {10,20,30}
print(s)

print(200 in s)  # boolean: True/False
print(20 in s)

print(10 not in s)
print(10 in s)
```

### 10. Set Operations(Most Important point)

Python sets support mathematical operations


1. union
2. intersection
3. difference
4. symmetric_difference

```python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
```

### Union: Unique value

##### 1.union()

##### symbol- pipe(|)
#### method - union()

```python
C = A.union(B)
print(C)
```

```python
print(A | B)
```

#### 2. Intersection

Common elemnets

Operator: & (Ampersand)

```python
A = {1,2,3}
B = {3,4,5}

C = A.intersection(B)  # intersection()
print(C)
```

```python
print(A & B)  # symbol
```

##### 3.difference
##### difference() symbol: -

```python
A = {1,2,3}
B = {3,4,5}

C = A.difference(B)  # A-B
print(C)

C1 = B.difference(A) # B-A
print(C1)
```

```python
# symbol: -

print(A-B)
```

```python
print(B-A)
```

##### 4. symmetric_difference

Elements in either set but not both

Operator: ^ [Carot]

Method : symmetric_difference()

```python
A = {1,2,3}
B = {3,4,5}

C = A.symmetric_difference(B)
print(C)
```

```python
print(A^B)
```

### 11.Set Comparison Operations

subset()

Check if set is subset().

```python
A = {1,2}
B = {1,2,3}

print(A.issubset(B))
```

```python
print(B.issubset(A))
```

superset()

```python
A = {1,2}
B = {1,2,3}

print(B.issuperset(A))
```

```python
print(A.issuperset(B))
```

## disjoint

Check it not common elements

```python
A = {1,2}
B = {3,4}

print(A.isdisjoint(B))
```

```python
A = {1,2,3}
B = {3,4}

print(A.isdisjoint(B))
```

### 12. Frozen Set

A `frozenset` is an immutable version of set

Once created, it cannot be modified.

Example:


```python
fs = frozenset([1,2,3])
print(fs)
```

```python
fs
```

```python
fs.add(10)
```

```python
s = {1,2,3}
s.add(10)
print(s)
```

### 13. Set Comprehension

Similar to list Comprehension.

```python
s = {x*x for x in range(5)}  # perform loop/iterate inside the set 
print(s)
```

### 14. Practical Example

Remove duplicates values from list

```python
numbers = [1,2,2,3,4,4,5]  # list
print(numbers)
unique = set(numbers)  # set
print(unique)
```

### 15. Convert Set to List


```python

s = {1,2,3}
print(s)
print(type(s))


lst = list(s)  # convert set to list
print(lst)
print(type(lst))
```

### 17. Performance Advantage of Sets

Sets are hash tables, so operations are very fast.

|Operation|Complexity|
|----|----|
|add|O(1)|
|remove|O(1)|
|search|O(1)|

Better than lists for membership testing

### 18. When to Use Sets

Use sets when:

* ✅ Remove duplicates
* ✅ Fast Searching
* ✅ Mathematical Operations: union, intersection, difference, symmetric_difference,frozenset, superset, subset,disjoint
* ✅ Membership testing


Example:
* Unique user IDs
* Unique email addresses
* Tag systems
* Data Filtering
