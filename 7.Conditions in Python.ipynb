## 1. What is a Conditional Statement?

A `conditional statement` allows a program to make decision based conditions(True/False).


In Python, We use:
* if
* elif
* else

## 2. Bisc if Statement

✅ Syntax:

if condition:
    
    ## Code execute if condition is True

```python
age = 19

if age>=18:
    print("You can vote")
```

* Key points:
* Condition must return True/False
* Indendation is mandatory
* No parehenese required(unlike c/java)

#### 3. if -else statment

```python
age = 15

if age>=18:
    print("You can vote")
else:
    print("You cannot vote")
```

### Key Points:

** else runs when condition is False

** Only one else allowed per if

### 5. Nested if Statements

```python
age = int(input("Enter your Age:"))
citizen = True
# print(citizen)
if age>=18:
    if citizen:
        print("Elegible to vote")
    else:
        print("Not Elegible")
else:
    print("Underage")
```

### 10. Truthy and Falsy Values

----
Falsy Values:

* False
* None
* 0
* "" (empty string)
* [],{},()


```python
name = ""

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = False

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = None

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = 0

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = []

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = ()

if name:
    print("Name exits")
else:
    print("Name not exits")
```

```python
name = {}

if name:
    print("Name exits")
else:
    print("Name not exits")
```

### 12. Using in and not in

```python
fruits = ["apple","banana"]

if 'Banana'.lower() in fruits:
    print("Available")
else:
    print("Fruits not available")
```

```python
fruits = ["apple","banana"]

if 'Banana'.lower() not in fruits:
    print("Available")
else:
    print("Fruits not available")
```

```python
fruits = ["apple","banana"]

if 'Banana' in fruits:
    print("Available")
else:
    print("Fruits not available")
```

## 13. Advanced:  contions(Case)



```python
day = 7

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid")
```
