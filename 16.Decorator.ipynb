### 16. Decorator in Python

-----

### Q. What is Decorator? 

A `decorator` is a funciton that modifies another function's behavior without changing its code.


📌 Define 
> A decorator adds extra feature to a function.

### Why Decorators are Needed 

❌Without Decorators 

if you want to:

* Log function calls
* Check user authentication
* Measure execution time

You would have to repeat the same code in every function.

✅️With Decorator 

* Write the logic once.
* Resue it many times.
* Cleaner and maintainable code.

### Real-Life Analogy

Gift = Original function

wrapper = Decorator

Gift remians same, wrapper adds value

### Function are First-Class object [Concept]

### basic decorator Structure

```python
def decorator(func):  # func = say_hello
    def wrapper():
        print("Before Function")
        func()  # x()  # say_hello()   -  Hello world
        print("After Function")
    return wrapper
```

```python
def say_hello():
    print("Hello world")
say_hello()
```

## Applying a Decorator(Manual way)

```python
decorator(say_hello)
```

### Using @ syntax [Recommended]

```python
@decorator
def say_hello():
    print("Hello world")
say_hello()
```

### Decorator with Arguments

```python
def decorator(func):  # func = greet
    def wrapper(name):  ### Gauri
        print("Congratulation!!!")
        func(name)  ### greet("Gauri")
    return wrapper
```

```python
@decorator
def greet(name):
    print("Hello", name)
greet("Gauri")
```

```python
import time
def timer(func):  
    def wrapper(*args, **kwargs): 
        print("Function Started")
        start = time.time()  # starting time
        result = func(*args, **kwargs)
        end = time.time()  # ending time
        
        print("Execution Time:", end-start)
        print("Function ended")
        return result
    return wrapper
```

```python
@timer
def demo():
    for i in range(1000000000):
        pass
demo()
```
