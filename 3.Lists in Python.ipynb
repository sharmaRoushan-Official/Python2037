C). extend(): Adds multiple elements at a time.

```python
a = [1,2,3]
print(a)
print(type(a))
```

```python
a = [1,2,3]

print(a)
print(type(a))
```

### 7. Removing elements from the list

A). remove()  --> Removes by value

B). pop(n)  --> n is index value, Removes by index value[return removed element]

C). pop() without index --> Removes last element 

D). clear()  --> Removes all elements

E). del keyword

------

A). remove() --> Removes by value

```python
nums = [10,20,30,40]

nums.remove(20)  # 20 will be removed from list
```

```python
print(nums)
```

B). pop(n) --> n is index value, Removes by index value[return removed element]

```python
nums = [10,20,30,40]
print(nums)

x = nums.pop(1)  # 1 is a index value of list 
print(x)
```

```python
print(nums)
```

C). pop() without index --> Removes last element

```python
nums = [10,20,30,40]
print(nums)

nums.pop()
```

```python
print(nums)
```

D). clear() --> Removes all elements

```python
nums = [10,20,30,40]
print(nums)

nums.clear()

print(nums)
```

E). del keyword

```python
nums = [10,20,30,40,50,60]
print(nums)
```

```python
del nums[1]  # 1 is the index value of element
```

```python
nums
```

```python
del(nums[0])
print(nums)
```

### 8. List Methods 
----

|Method|Work|
|----|---|
|append(x)|Add item at end|
|extend(iterable)| Add multiple items|
|insert(i,x)| insert at index position|
|remove(x)| Remove x value|
|pop(n)| pop element n and n is index value|
|pop()| pop last element|
|clear()| removes all elements|
|del(n) | delete item of n position|
|index(x)| Return index of value x|
|count(x)| Counts occurrence of x value|
|sort()| Sort list: asc|
|reverse()| Reverse the list|
|copy()| Returns shallow copy|

## 9. index(x) and Count(x)

```python
nums = [10,20,30,40,50]
print(nums)
print(nums.index(20))
print(nums.index(50))
# print(nums.index(60)) ValueError 

```

```python
# count()

nums = [10,20,30,40,50,20,40,30,30]

print(nums.count(10))
print(nums.count(20))
print(nums.count(30))
print(nums.count(40))
print(nums.count(60))  # 0 no element 
```

## 10. Sorting list

Ascening sort()

```python
nums = [10,20,30,40,50,20,40,30,30]

print("Original List:",nums)

nums.sort()
print(nums)
```

```python
# Reverse() --> Descending

nums = [10,20,30,40,50,20,40,30,30]

print("Original List:",nums)

nums.sort()
print(nums)
```

```python
num1 = nums[::-1]  # slicing method
print(num1)
```

```python
nums = [10,20,30,40,50,20,40,30,30]

print("Original List:",nums)
```

```python
nums.sort(reverse=True)  # In-built method 
print(nums)
```

```python
x = ["Arnav","Trupti","Pankaj",10,20]
print(x)
```

```python
y = x.copy()
print(y)
```
