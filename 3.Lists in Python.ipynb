## 3. Lists in Python

* A `List` in Python is a mutable(Can be changes), Ordered, Indexed collection that can store mutiple values.
* including(mixed data types).
* Lists are one of most used data types in Python.

## 1. List Defintion
----

syntax:

my_list = [1,2,3,4]


```python
my_list = [1,2,3,4]
print(my_list)
```

```python
data = [10,20,30]
print(data)
```

### 2. Features of List
-----

|Feature|Explanation|
|---|---|
|Ordered| Items maintain insertion order|
|Indexed| Each element has an index|
|Mutable| Can modify after creation|
|Allows duplicates| Same values can repeat|
|Mixed types allowed| int, float, str,bool,object, tuple, list, set, dictionary|

```python
x = [10,25.36,"Python",True, False, [10,20],(10,20),{1,2,3},{'a':10,"b":90}]
print(x)
print(type(x))
```

## 3. List Indexing 

### Positive Indexing


### Negative Indexing

```python
x = [10,25.36,"Python",True, False, [10,20],(10,20),{1,2,3},{'a':10,"b":90}]
print(x)
print(type(x))
```

```python
print(x[0])
print(type(x[0]))
print(x[1])
print(type(x[1]))
print(x[2])
print(type(x[2]))
print(x[3])
print(type(x[3]))
print(x[4])
print(type(x[4]))
print(x[5])
print(type(x[5]))
print(x[6])
print(type(x[6]))
print(x[7])
print(type(x[7]))
print(x[8])
print(type(x[8]))
# print(x[9]) IndexError: list index out of range
```

## Negative Indexing 

```python
## Negative Indexing 

x = [10,25.36,"Python",True, False, [10,20],(10,20),{1,2,3},{'a':10,"b":90}]
print(x)
print(type(x))


print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
print(x[-7])
print(x[-8])
print(x[-9])
# print(x[-10]) IndexError: list index out of range
```

## 4. Slicing of list

```python
x = [10,25.36,"Python",True, False, [10,20],(10,20),{1,2,3},{'a':10,"b":90}]
print(x)
print(type(x))
```

```python
x[0:5]
```

```python
x[5:6]
```

```python
x[7:]
```

### 5. Striding of List

```python
x = [10,25.36,"Python",True, False, [10,20],(10,20),{1,2,3},{'a':10,"b":90}]
print(x)
print(type(x))
```

```python
x[0:9:2]
```

```python
x[::2]
```

```python
x[::3]
```
