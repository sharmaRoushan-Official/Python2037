```python
#1. Create a list of 5 numbers and print it.


#2. Write a program to find the length of a list.


#3. Write a program to add an element to the end of a list.


#4. Write a program to insert an element at a specific position in a list.


#5. Write a program to remove an element from a list.


#6. Write a program to find the largest number in a list.


#7. Write a program to find the smallest number in a list.


#8. Write a program to sort a list in ascending order.


#9. Write a program to count how many times a specific element appears in a list.


#10. Write a program to reverse a list.


```

#### 1. Create a list of 5 numbers and print it.


```python
numbers=[10,20,30,40,50]
print(numbers)
```

#### 2. Write a program to find the length of a list.


```python
numbers=[10,20,30,40,50]
print(len(numbers))
```

#### 3. Write a program to add an element to the end of a list.


```python
numbers=[10,20,30,40,50]
numbers.append(60)
print(numbers)
```

#### 4. Write a program to insert an element at a specific position in a list.


```python
numbers=[10,20,40,50]
numbers.insert(2,30)
print(numbers)
```

#### 5. Write a program to remove an element from a list.



```python
numbers=[10,20,30,40,50]
numbers.remove(30)
print(numbers)
```

#### 6. Write a program to find the largest number in a list.


```python
numbers=[10,20,30,40,50]
print(max(numbers))
```

#### 7. Write a program to find the smallest number in a list.


```python
numbers=[10,20,30,40,50]
print(min(numbers))
```

#### 8. Write a program to sort a list in ascending order.


```python
numbers=[20,60,10,40,50]
numbers.sort()
print(numbers)
```

#### 9. Write a program to count how many times a specific element


```python
numbers=[20,60,10,10,50]
print(numbers.count(10))
```

#### 10. Write a program to reverse a list.

```python
numbers=[20,60,10,10,50]
numbers.reverse()
print(numbers)
```
