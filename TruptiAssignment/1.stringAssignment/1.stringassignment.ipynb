#### 5. write a program to count the numbers of vowels in a string

```python
s=("trupti")
count=0
for ch in s.lower():
    if ch in "aeiou":
        count+=1
        print(count)
```

#### 6. write a program to reverse a string

```python
s6=("Trupti","Vedika","Ankita")
print(s6[::-1])
   
```

#### 7. write a program to check whether a given string is a palindrome

```python
s=("trupti")
if s==s[::-1]:
    print("palindrome")
else:
    print("not palindrome")
```

#### 8. write a program to count the occurences of a specific character in a string


```python
s9=("trupti","vedika","trupti")
s=("trupti")
print(s9.count(s))
```

#### 9. write a program to replace a word in a string with another word
