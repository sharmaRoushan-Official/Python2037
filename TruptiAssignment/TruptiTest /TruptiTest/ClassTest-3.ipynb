# 3️⃣ Condition

## Write a program to input a year and check whether it is a leap year or not.


```python
year=int(input("enter year:"))
if year%4==0:
    print("Leap year")
else:
    print("not leap year")
```

# 4️⃣ For Loop

## Write a program to print the multiplication table of a given number from 1 to 10.


```python
for i in range(1,11):
    print(f"--------Table of{i}-------")
    for j in range(1,11):
        print(i,"X",j,"=",i*j)
print()
```

# 5️⃣ For Loop

## Write a program to find the sum of all even numbers between 1 and N.


```python
n=1
s=0
for i in range(1,n+1):
    s+=1
    print(s,end=" ")
```

## 6️⃣ For Loop

## Write a program to find the factorial of a number using a "for" loop.


```python
n=10
fact=1
for i in range(1,n+1):
    fact*=i
    print(fact,end=" ")
```

# 7️⃣ While Loop

## Write a program to reverse a given number using a "while" loop.
Example: "12345 → 54321"


```python
num=12345
rev=0
while num>0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
print(rev)
```

# 8️⃣ While Loop

* Method: "display()"
