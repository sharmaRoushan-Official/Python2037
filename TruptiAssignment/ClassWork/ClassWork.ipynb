## 5. Count the total number of words in a file.


```python
fileName=(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt")
with open(fileName,"r") as file:
    data=file.read()
    words=data.split()
    countwords=0
    for word in words:
        countwords+=1
    print("total words in this file:",countwords)
    
```

## 6. Count the total number of characters in a file.


```python
fileName=(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt")
with open(fileName,"r") as file:
    countchar=0
    space=0
    for data in  file.read():
        if data==" ":
            space+=1
        countchar+=1
    print("total character in this file:",countchar)
    print("total space in this file:",space)
    print("total number of character without space:",countchar-space)
        
```

## 10. Read a file and print each word on a separate line.


```python
fileName=(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt")
with open(fileName,"r") as file:
    data=file.read()
    word=""
    for ch in data:
        if ch ==" " or ch=="\n" or ch=="\t":
            if word !=" ":
                print(word)
                word=""
        else:
            word+=ch
```

## 11. Count the number of vowels present in a text file.


```python
fileName=(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt")
with open(fileName,"r") as file:
    content=file.read()
    vowels = "aeiou"
    count = 0
for char in content:
    if char in vowels:
        count +=1
print( count)

```

## 12. Count the number of uppercase and lowercase letters in a file.


```python
fileName=(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt")
upperchar=0
uppercharacter=[]
lowerchar=0
lowercharacter=[]
with open(fileName,"r") as file:
    for ch in file.read():
        if ch.isupper():
            uppercharacter.append(ch)
            upperchar+=1
        else:
            if ch==" " or ch=="\n" or ch=="\t":
                continue
            lowercharacter.append(ch)
            lowerchar+=1
print("upper character is:",uppercharacter,"=",upperchar)
print("lower character is:",lowercharacter,"=",lowerchar)            

   
        
```
