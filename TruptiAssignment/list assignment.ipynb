## list question

#### create list of 5 numbers and print it python

```python
numbers=[10,20,30,40,50]
print(numbers)
```

#### find the length of a list python

```python
numbers=[10,20,30,40,50]
print(len(numbers))
```

#### Add  an element to the end of a list python

```python
numbers=[10,20,30,40,50]
numbers.append(60)
print(numbers)
```

#### insert an element at a specific position python

```python
numbers=[10,20,40,50]
numbers.insert(2,30)
print(numbers)
```

#### remove an element from a list python

```python
numbers=[10,20,30,40,50]
numbers.remove(30)
print(numbers)
```

#### find the largest number in a list python

```python
numbers=[10,20,30,40,50]
print(max(numbers))
```

#### find the smallest number in list python

```python
numbers=[10,20,30,40,50]
print(min(numbers))
```

#### sort a list in ascending order python

```python
numbers=[20,60,10,40,50]
numbers.sort()
print(numbers)
```

#### count occurences of an element python

```python
numbers=[20,60,10,10,50]
print(numbers.count(10))
```

#### reverse a list python

```python
numbers=[20,60,10,10,50]
numbers.reverse()
print(numbers)
```
