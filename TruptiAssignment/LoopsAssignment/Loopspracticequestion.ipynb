## 29. Find the GCD of two numbers using a loop.


```python
# for loop                  greatest common division


a=int(input("enter first number:"))
b=int(input("enter second number:"))
gcd=1
for i in range(1,min(a,b)+1):
    if a%i==0 and b%i==0:
        gcd=i
print(gcd)
```

## 1. Count the number of even and odd numbers in a given list using a for loop.


```python
# for loop



list=[1,2,3,4,5,6,7]
even=0
odd=0
for i in list:
    if i%2==0:
        even+=1
    else:
        odd+=1
print("even=",even)
print("odd=",odd)
```

```python
# while loop



list=[1,2,3,4,5,6,7]
even=0
odd=0
i=0
while i<len(list):
    if list[i]%2==0:
        even+=1
    else:
        odd+=1
    i+=1
print("even=",even)
print("odd=",odd)
```

## 2. Find the second largest number in a list without using sort() or max().


```python
# for loop


list=[10,20,30,40,]
largest=second=-9999
for num in list:
    if num>largest:
        second=largest
        largest=num
    elif num>second and num!=largest:
        second=num
print(second)
```

```python
# while loop



list=[10,20,30,40,]
largest=second=-9999
i=0
while i<len(list):
    num=list[i]
    if num>largest:
        second=largest
        largest=num
    elif num>second and num!=largest:
        second=num
    i+=1
print(second)
```

## 10. Check whether a given number is an Armstrong number.


```python
# for loop

num=153
temp=str(num)
sum1=0
for digit in temp:
    sum1+=int(digit)**len(temp)
if sum1==num:
    print("armstrong")
else:
    print("not armstrong")
```

```python
# while loop

num=153
temp=num
sum1=0
while temp>0:
    digit=temp%10
    sum1+=digit**3
    temp//= 10
if sum1==num:
    print("armstrong")
else:
    print("not armstrong")
```

## 20. Given a dictionary of student names and marks:



Find the topper.

Calculate the average marks.

Print students scoring above average.

Count students scoring below 70.


```python
# for loop

student={
    "A":80,
    "B":95,
    "C":60,
    "D":70
}
topper=max(student,key=student.get)
print("topper=",topper)
total=0
for i in student:
    total+=student[i]
avg=total/len(student)
print("Average=",avg)
for i in student:
    if student[i]>avg:
        print("Above Average=",i)
count=0
for i in student:
    if student[i]<70:
        count+=1
print("Below 70=",count)
    
```

```python
# while loop


student={
    "A":80,
    "B":95,
    "C":60,
    "D":70
}
keys=list(student.keys())
i=0
total=0
while i<len(keys):
    total+=student[keys[i]]
    i+=1
avg=total/len(keys)
print("Average=",avg)
    
```

## 21. Print Pascal's Triangle.


```python
# for loop

n=5
for  i in range(n):
    num=1
    for j in range (i+1):
        print(num,end=" ")
        num=num*(i-j)//(j+1)
    print()
```

```python
# while loop


n=5
i=0
while i<n:
    num=1
    j=0
    while j<=i:
        print(num,end=" ")
        num=num*(i-j)//(j+1)
        j+=1
    print()
    i+=1
```

## 23. Print all perfect numbers between 1 and 1000.


```python
# for loop

for num in range(1,1001):
    total=0
    for i in range(1,num):
        if num%i==0:
            total+=i
    if total==num:
                print(num)
```

```python
# while loop

num=1
while num<=1000:
    total=0
    i=1
    while i<num:
        if num%i==0:
            total+=i
        i+=1
    if total==num:
        print(num)
    num+=1
```

## 24. Find all pairs in a list whose sum equals a target value.


```python
# for loop

list=[1,2,3,4,5]
target=6
for i in range(len(list)):
    for j in range(i+1,len(list)):
        if list[i]+list[j]==target:
            print(list[i],list[j])
```

```python
# while loop

list=[1,2,3,4,5]
target=6
i=0
while i<len(list):
    j=i+1
    while j<len(list):
        if list[i]+list[j]==target:
            print(list[i],list[j])
        j+=1
    i+=1
```

## 25. Implement Selection Sort using for loops.


```python
# for loop


list=[64,25,12,22,11]
for i in range(len(list)):
    min_index=i
    for j in range(i+1,len(list)):
        if list[j]<list[min_index]:
            min_index=j
    list[i],list[min_index]=list[min_index],list[i]
print(list)
```

```python
# while loop

list=[64,25,12,22,11]
i=0
while i<len(list):
    min_index=i
    j=i+1
    while j<len(list):
        if list[j]<list[min_index]:
            min_index=j
        j+=1
    list[i],list[min_index]=list[min_index],list[i]
    i+=1
print(list)
```

## 26. Implement Bubble Sort using nested for loops.


```python
# for loop

list=[64,25,12,22,11]
for i in range(len(list)):
    for j in range(0,len(list)-i-1):
        if list[j]>list[j+1]:
            list[j],list[j+1]=list[j+1],list[j]
print(list)
```

```python
# while loop

list=[64,25,12,22,11]
i=0
while i<len(list):
    j=0
    while j<len(list)-i-1:
        if list[j]>list[j+1]:
            list[j],list[j+1]=list[j+1],list[j]
        j+=1
    i+=1
print(list)
```
