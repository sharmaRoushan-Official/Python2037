## 21. Read a file and find the frequency of each word.


```python
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    content = file.read().lower()
words = content.split()
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1
for word, count in frequency.items():
    print(word, ":", count)
```

## 22. Read a file and display the top 10 most frequently used words.


```python
from collections import Counter
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    content = file.read().lower()
words = content.split()
frequency = Counter(words)
top10 = frequency.most_common(10)
for word, count in top10:
    print(word, ":", count)

```

## 27. Read a CSV file and calculate the average of a numeric column.


```python
import csv
total=0
count=0
with open("data.csv","r") as file:
    reader=csv.DictReader(file)
    for row in reader:
        total=total+int(row["Age"])
        count=count+1
average=total/count
print("Average=",average)
```

## 30. Read a large file efficiently using chunks instead of loading the entire file into memory.

```python
with open("largefile.txt","r")as file:
    while True:
        chunk=file.read(1024)
        if not chunk:
            break
        print(chunk)
```

## Read a log file and count the number of ERROR, WARNING, and INFO messages.


```python
error = 0
warning = 0
info = 0
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt", "r") as file:
    for line in file:
        if "ERROR" in line:
            error += 1
        elif "WARNING" in line:
            warning += 1
        elif "INFO" in line:
            info += 1
print("ERROR:", error)
print("WARNING:", warning)
print("INFO:", info)
```

## Read a text file and find the longest word.


```python
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    content = file.read()
words = content.split()
longest_word = max(words, key=len)
print( longest_word)
```

## Read a file and remove duplicate lines.


```python
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    lines = file.readlines()
unique_lines = []
seen = set()
for line in lines:
    if line not in seen:
        unique_lines.append(line)
        seen.add(line)
for line in unique_lines:
    print(line, end="")
```

## Read a file and print all palindrome words.


```python
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    content = file.read().lower()
words = content.split()
for word in words:
    word = word.strip(".,!?")
    if word == word[::-1]:
        print(word)
```

## Read a file and count the occurrence of every alphabet (A–Z).

```python
from collections import Counter
with open(r"C:\\Users\\acer\\OneDrive\\Desktop\\NursaryRhyms.txt","r") as file:
    content = file.read().lower()
count = Counter()
for char in content:
    if char.isalpha():
        count[char] += 1
for letter in "abcdefghijklmnopqrstuvwxyz":
    print(letter, ":", count[letter])
```
