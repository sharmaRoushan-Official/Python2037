## 1️⃣7️⃣ Write all prime numbers between 1 and 100 into a file.


```python
with open("prime.txt", "w") as f:
    for num in range(1, 101):
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            f.write(str(num) + "\n")
```

```python
with open("result.txt", "w") as f:
    for i in range(10):
        name = input("Enter Name: ")
        eng = int(input("English Marks: "))
        math = int(input("Math Marks: "))
        sci = int(input("Science Marks: "))
        total = eng + math + sci
        per = total / 3
        f.write(name + ", ")
        f.write(str(total) + ", ")
        f.write(str(per) + "%\n")
```

## 3️⃣0️⃣ ⭐ Mini Project – Student Record Management System

Create a file-based Student Management System with the following menu:

1. Add Student
2. Add Multiple Students
3. Add Marks
4. Generate Result
5. Exit

Store all student information in a text file.


```python
 fileName = "student.txt"


def add_student():
    name = input("Enter Student Name: ")
    roll = input("Enter Roll Number: ")
    with open(fileName,'a') as f:
        f.write("=============================================\n")
        f.write(f"Student Name: {name}\nStudent Roll No.{roll}\n")
        f.write("=============================================\n")

def add_multiple_student():
    while True:
        add_student()
        choice = input("Add Another? (y/n): ")
        if choice.lower() != "y":
            break
      
while True:
    print("\n===== Student Management System =====")
    print('''
    \n1.Add Student
    \n2.Add Multiple Students
    \n3.Add Marks
    \n4.Generate Result
    \n5.Exit.
    ''')
    choice = input("Enter Choice: ")
    if choice == "1":
        add_student()
    elif choice == "2":
        add_multiple_student()
    elif choice == "5":
        break
```

## Write a table  into file from 1-10 "table.txt"

```python
with open("table.txt","w") as f:
    for i in range(1,11):
        f.write(f"============Table of{i}============\n")
        for j in range(1,11):
            f.write(f"{i}X{j}={i*j}\n")
        f.write("\n")
    f.write(f" End of the Table from 1 to {j}")
```

## Given sentence write in the file "sentence.txt" count total number of characters and words in the file.

### Sentence = A Quick brown fox jump over the lazy dog.

### output:
* Total number of words=9
*  Total number of characters=41

```python
with open("sentence.txt","w") as file:
    sentence = "A Quick brown fox jump over the lazy dog."
    totalChar=0
    words=sentence.split()
    wordCount=0
    for word in words:
        wordCount+=1
        for  char in word:
            totalChar+=1
    file.write(f"Sentence:{sentence} \n")
    file.write(f"Total words in your sentence:{wordCount} \n")
    file.write(f"Total characters in your sentence:{totalChar} \n")
```
