## 16. Write a function that accepts a list of integers and returns the sum of even numbers.


```python
def sum_even(list):
    a=0
    for i in  list:
        if i%2==0:
            a+=1
    return a
print(sum_even([1,2,3,4,5,6]))
```

## 17. Write a function that takes a sentence and returns the number of words.


```python
def words(sentence):
    count=1
    for ch in sentence:
        if ch==" ":
            count+=1
    return count
print(words("python is case sensative language"))
```

## 18. Write a function that takes a number and returns the sum of its digits.


```python
def sum(number):
    count=0
    while number:
        digit=number%10
        count=count+digit
        number=number//10
    return count
print(sum(3456))
```

## 27. Write a function to remove punctuation from a string.


```python
import string
def remove_punctuation(text):
    result=""
    for ch in text:
        if ch not in string.punctuation:
            result+=ch
    return result
print(remove_punctuation("Hello,python!"))
```

## 28. Write a function to calculate the compound interest.


```python
def compound_interest(p,r,t):
    return p*((1+r/100)**t)-p
print(compound_interest(10000,10,2))
```

## 29. Write a recursive function to calculate factorial.


```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

## 30. Write a recursive function to compute the nth Fibonacci number.


```python
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1)+fibonacci(n-2)
print(fibonacci(6))
```

## 34. Write a function that takes a list and returns a new list with only unique elements (preserving order).


```python
def unique_list(lst):
    result=[]
    for i  in lst:
        if i not in result:
            result.append(i)
    return result
print(unique_list([1,2,2,3,1,4]))
        
```

## 39. Write a function using map() to double all elements in a list.


```python
def double(lst):
    return list(map(lambda x: x * 2, lst))
print(double([1, 2, 3, 4]))
```

## 40. Write a function using filter() to get all odd numbers from a list.


```python
def odd_numbers(lst):
    return list(filter(lambda x: x % 2 != 0, lst))
print(odd_numbers([1, 2, 3, 4, 5, 6]))
```

## 41. Write a function using reduce() to find the product of all numbers in a list.


```python
from functools import reduce
def product(lst):
    return reduce(lambda x, y: x * y, lst)
print(product([2, 3, 4]))
```

## 44. Write a decorator function that logs the execution time of another function.


```python
import time
def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print("Execution Time:", end - start)
    return wrapper
@timer
def test():
    print("Running...")
test()
```

## 45. Write a recursive function to find the sum of all elements in a list.


```python
def list_sum(lst):
    if len(lst) == 0:
        return 0
    return lst[0] + list_sum(lst[1:])
print(list_sum([1, 2, 3, 4]))
```

## 48. Write a function that returns a list of even numbers using list comprehension.


```python
def even_numbers(n):
    return [i for i in range(n + 1) if i % 2 == 0]
print(even_numbers(10))
```

## 49. Write a function that takes a list of strings and returns a new list with all strings in uppercase.
