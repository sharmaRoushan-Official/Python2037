## tuple question:

#### create a tuple with 5 values python

```python
t=(10,20,30,40,50)
print(t)
```

#### find the length of a tuple python

```python
t=(10,20,30,40,50)
print(len(t))
```

#### access the third element python

```python
t=(10,20,30,40,50)
print(t[2])
```

#### check whether an element exists python

```python
t=(10,20,30,40,50)
if 30 in t:
    print("exists")
else:
    print("not exists")
```

#### count occurences of a value python

```python
t=(10,20,30,40,50)
print(t.count(10))
```

#### find index of an element python

```python
t=(10,20,30,40,50)
print(t.index(20))
```

#### concatenate two tuples python

```python
t1=(1,2,3)
t2=(4,5,6)
t3=t1+t2
print(t3)
```

#### convert tuple into list python

```python
t=(10,20,30,40,50)
t1=list(t)
print(t1)
```

#### find maximum and minimum values python

```python
t=(10,20,30,40,50)
print(max(t))
print(min(t))
```

#### iterate through tuple elements python

```python
t=(10,20,30,40,50)
for i in t:
    print(i)
```

### find common elements between two lists python

```python
list1=[1,2,3,4,5]
list2=[3,4,5,6,7]
common=[]
for i in list1:
    if i in list2:
        common.append(i)
        print(common)
```
