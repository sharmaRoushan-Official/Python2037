## 9. Functions in Python

### 1.What is a Fucntion?

A `Function` is a reusable block of code that perform a specific task.

### Why We use Function?

 * Avoid code repetition
 * Make code readable
 * Improve modularity
 * Easier debugging
 * Resuability

### 4. Function with Return Value

✅ return Statement 

Return Value back to caller.

```python
def add(a,b):  # a, b is a parameter of function add
    return a+b  # return the value after terminate function
    
result = add(10,20)  # 10,20 argument
print(result)
```

### 5. Scope of Variables

#### 1. Local Variable

#### 2. Global Variable


### 1. Local Variable


```python
def test():
    x = 10  # x is local variable can be accessed only function block
    print(x)
test()
```

```python
print(x)
```

### 2. Global Variable

Access inside and outside both scope of the function

```python
x = 100

def text():
    print(x)  # x is accessed inside the function block
text()
```

```python
print(x)  # Accessed outside the function block
```

### global Keyword




```python
x = 5  # x-global variable

def change():
    x = 20  # x-local variable
    print(x) # 20
change()
print(x)  # 5 --> Global variable 
```

global keyword

```python
def change():
    x1 = 5 # local variable
    print(x1)
change()
```

```python
print(x1)
```

```python
def change():
    global x1  # x1 is global variable created
    x1 = 5 # local variable
    print(x1)
change()
```

```python
print(x1)  # local to global variable accessed.
```

```python
## 21. Write a function to convert Celsius to Fahrenheit.


# (°C*9/5)+32

def CelsiusToFahrenheit():
    celsius = float(input("Enter °C value: "))
    faherenheit = (celsius*9/5)+32
    return f"{faherenheit} °F"
CelsiusToFahrenheit()
```

```python
## 21. Write a function to convert Fahrenheit to Celsius .

# (32°F − 32) × 5/9 = 0°C

def fahrenheitToCelsius():
    fahrenheit = float(input("Enter fahrenheit Value: "))
    celsius = (fahrenheit -32)* 5/9
    return f"{celsius} °F"

fahrenheitToCelsius()
```

```python
## 24. Write a function that takes a string and counts uppercase and lowercase letters.


def filterChracter():
    str1 = "A Quick Brown Fox Jumpy Over The Lazy Dog"
    
    upperStr = []
    lowerStr = []
    space = []
    for i in str1:
        if ord(i)>64 and ord(i)<91:
            # print(i)
            upperStr.append(i)
        elif ord(i) == 32:
            space.append(i)
        else:
            lowerStr.append(i)
        # print(i)
    print("Upper Character is:",upperStr)
    print("Space Character is:",space)
    print("Lower Chracter is:",lowerStr)
filterChracter()
```

```python
filterChracter()
```

```python
# 25. Write a function to find the common elements between two lists.

def commonElementFind():
    list1 = ['apple',10,20,20,"Banana"]
    
    list2 = [10,20,100,45,'apple',"Banana"]
    
    commonElement = []
    
    for i in list1:
        for j in list2:
            if i == j:
                if i not in commonElement:
                    commonElement.append(i)
        # print(i)
    print(commonElement)
commonElementFind()
```
