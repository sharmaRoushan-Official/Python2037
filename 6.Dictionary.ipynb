## 6. Dictionary in Python

---

1. What is Dictionary in Python?

A `dictionary` is a collection of key:values pairs.

* Each key is unique.
* Each key maps to a value.
* Dictionary are mutable(unhachable/changeable)
* They are written using curly braces {}.



```python
d1 = {}
print(d1)
print(type(d1))
```

```python
x1 = set()
print(x1)
print(type(x1))
```

```python
student = {
    "name":"Arnav",
    "age":17,
    "course":"Python"
}
print(student)
print(type(student))
```

name,age,course ---> Keys

"Arnav",17,"Python" ---> Values

### 14. Copy dictionary

Using copy()  shallow copy()

```python
data1 = {
    "name":"Trupti",
    "age":99,
    "course":"Python"
}
print(data1)
```

```python
data2 = data1.copy()
print(data2)
```

```python
data1['address'] = "Pune"
print(data1)
```

```python
print(data2)
```

```python
### Deep copy:

data1 = {
    "name":"Trupti",
    "age":99,
    "course":"Python"
}
print(data1)
```

```python
data2 = data1
print(data2)
```

```python
data2["address"] = "Pune"
print(data2)
```

```python
print(data1)
```
