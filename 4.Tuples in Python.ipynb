### 4. Tuples in Python
-----

#### 1.What is a Tuples?

A `tuple` is a collection data type in Python used to store multiple values in single variable.

## Key characteristics:

* Ordered collection
* Immutable (Cannot be changed after creation)
* Allows duplication values
* Can store different data types
* Faster than lists
* Written using round brackets()


```python
numbers = (10,20,30,40)
print(numbers)
```

```python
print(type(numbers))
```

## 2.Creating Tuples

#### Method 1 - Uinsg Parentheses


```python
fruits = ("apple","banana","mango")
print(fruits)
```

```python
print(type(fruits))
```

### Method -2 - Without Parentheses

Parentheses are optional

```python
colors = "red","green","blue"
print(colors)
print(type(colors))
```

### Method 3- Using tuple()   Constructor

```python
numbers = tuple([1,2,3,4,5,6])
print(numbers)
print(type(numbers))
```

### 3. Tuple with Different Data Types

Tuples can store multiple data types.
    

```python
data = ("john",25,75.5,True,False)
print(data)
print(type(data))
```

### 4. Nested Tuples : Tuples inside Tuples

A tuple inside another tuple

```python
nested_tuple = ((1,2),(3,4),(4,5))
print(nested_tuple)
print(type(nested_tuple))
```

## 6. Slicing of Elements:

```python
fruits[0:3:2]
```

```python
fruits[0:3:3]
```

### 7. Tuple Immutablability

Tuples cannot be modified after creating


```python
numbers = (10,20,30)
print(numbers)
```

```python
numbers[0] = 100
numbers
```

## 8. Changing Tuple Elements:

Since tuples are immutable, convert to list

```python
numbers = (10,20,30) 
print(numbers)  # tuple
print(type(numbers))  # immutable 

temp = list(numbers)   # tuple element converted into list data type
print(temp)
print(type(temp))


temp[0] = 100   # mutable
print(temp)
```

### 9. Loop Through Tuples

Using for loop

```python
fruits = ("apple","banana","mango")
print(fruits,type(fruits))


# traverse: visit all elemnets at least one time.
for i in fruits:
    print(i)
```

## 11. Tuple Packing

```python
data  = ("John",25,"Engineer")
print(data)
```

## 12. Tuple Unpacking

Extract tuple values into variables

```python
data = ('John',25,"Engineer")
print(data)

name,age,job = data

print(name)
print(age)
print(job)
```

### 13. Extended Unpacking

Using *operator (Asterisk) asterisk 


```python
numbers = (10,20,30,40,50)
print(numbers)
```

```python
a,b,c,d,e = numbers
print(a,b,c,d,e)
```

```python
a,*b,c = numbers
print(a)
print(b)
print(c)
```

### 17. Tuples Methods

Tuples have only one built-in methods

1.count()

Count occurrence

```python
numbers = (1,2,3,2,4,2)
print(numbers.count(2))
```

2.index()

Returns index of first occurrence.

```python
numbers = (10,20,30,30,40)
print(numbers)
```

```python
print(numbers.index(10))
print(numbers.index(20))
print(numbers.index(30)) # first occurrence
print(numbers.index(40))
```

### 20. Advantages of Tuples
--
* 1. Faster than lists
  2. Immutable(safe from accidental changes)
  3. Can be used as dictionary keys
  4. Consumes less memory

### 21.Tuple vs List

|Feature|Tuple|List|
|---|---|---|
|Brackets|()|[]|
|Mutable|No|Yes|
|Speed|Faster|Slower|
|Memory|Less|More|

```python
tuple_data = (1,2,3)
print(tuple_data)

list_data = [1,2,3]
print(list_data)
```

### 23. When to Use Tuple:
---
Use tuples when:

* Data should not change
* Fixed records(coordinate, database row)
* Returning multiple values from functions
