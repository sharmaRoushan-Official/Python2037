## Variables and Expression in Python   


###  1. Introduction to Varibales

A `variable` is a name used to store data in memory. [RAM-Random Access Memory]

A variable as a labelled container that hold information.

Example 

```python
name = "Trupti"
age = 25
salary = 45000.50
print(name)
print(age)
print(salary)
```

Here:

* `name` stores text
* `age` stores an integer
* `salary` a decimal value

## 2. Why Variables are Important
---
Variables help:

* Stores user input
* Perform calculations
* Reuse values
* Build logic in Program
* Meke code readable and manageable

### 3. Syntax of Varibles

variable_name = value

```python
x = 10
print(x)
```

* `x` -> Variable Name
* `=` -> Assignemnt operator
* `10` -> assigned value

## 4. Types of Variables in Python
Python automatically identifies data types

A. Integer(int)



B. Float(float)


C. String(str)


D. Boolean(bool)

```python
# A. Integer(int)
age = 30
print(age,type(age))

# B. Float(float)
price = 99.99
print(price,type(price))

# C. String(str)
city = "NOIDA"
print(city, type(city))

# D. Boolean(bool)
is_active = True

print(is_active, type(is_active))
```

### 5. Dynamic Typing in Python

A Python is dynamically typed language.

A variable can change its datatype.

```python
x = 10
print(x,type(x))

# Now `x` stores a integer
```

```python
x = "hello world"
print(x,type(x))
```

```python
# Now `x` stores a string
```

### Rules for Creating Variables in Python

#### 1. Variable Name Must starts with 
* ✅ Letter (a-z, A-Z)
* ✅ Uderscore(_)

```python
name = "Raj"
_age =30
city1 = "Texas"

print(name)
print(_age)
print(city1)
```

### 2. Variable Name Cannot start with Number

❌ Invalid

```python
1name = "Raj"
```

✅ Correct


```python
name1 = "Raj"
print(name1)
```

### 3. No Spaces Allowed
❌ Invalid

```python
first name = "Raj"
print(first name)
```

✅ Correct

```python
first_name = "Raj"
print(first_name)
```

## 4.Special Symbol is Not allowed
Except underscore `_`

❌ Invalid

```python
price@ = 99.99
print(price@)
```

✅ Correct

```python
price_value = 99.99
print(price_value)
```
