### type() function is used to check their data type 

```python
# 1.String
print("Hello world")
print(type("Hello world"))
print('hello world')
print(type('hello world'))
print('100')
print(type('100'))
print("1565641.0")
print(type("1565641.0"))


# 2. Integer
print(100)
print(type(100))
print(1000)
print(type(1000))
print(1234)
print(type(1234))


# 3. Float
print(100.0)
print(type(100.0))
print(1000.123)
print(type(1000.123))
print(4567.25)
print(type(4567.25))


# 4. Boolean
print(True)
print(type(True))  # 1
print(False)
print(type(False))  # 0
```
