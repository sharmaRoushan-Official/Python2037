### 12.Reading Files in Python

##### What is File Handling?

`File handling` allows python program to read data files and write data to files permanently.

📌 Why we read files:

* Stored data permanenlty
* Read logs, reports, CSV[Comma Seperated Value]
* Process large data
* Avoid manual input

### Types of Files

1. Text Files (.txt, .csv, .log)
2. Binary files (.jpg,.pdf, .exe)

> Here we focus on reading text files.

----
## ╰┈➤ Steps to Read a File in Python

1. Open the file
2. Read the file
3. Close the File

##### ╰┈➤ Opening a File

`file = open("fileName","r")`

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
# print(fileName)
file = open(fileName,'r')  # step 1
content = file.read()  # step 2
print(content)
file.close()  # step 3
```

|Mode|Meaning|
|----|---|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`|Read & write|

### Method 1: read() --> Read Entire File


```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,'r')
content = file.read()
print(content)
file.close()
```

----
Notes:

* Reads whole file at once.
* No suitable for very large files.

### Method 2: readline() -> Read One Line



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file  = open(fileName,'r')

line = file.readline()
print(line)
print(line)
print(line)
print(line)
file.close()
```

📌 Reads only one line at a time.

#### Method 3 : readlines() --> Read all lines as list



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file  = open(fileName,'r')

content = file.readlines()

print(content)

file.close()
```

### 6. read(n) method 

Reads n chracters only.

Example

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,'r')
content = file.read(10)  # 10 characters
content1 = file.read(20)  # next 20 characters
print(content)
print(content1) 
file.close()
```

### 7. readline() Method

### 8. readlines() Method

Reads all lines into a list.

Example:

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file  = open(fileName,'r')

content = file.readlines()

print(content)

file.close()
```

## 9. Readline File Using Loop. 

Best method for large files.

Example:

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
file  = open(fileName,"r")

for line in file:
    print(line)

file.close()
```
