### 12.Reading Files in Python

##### What is File Handling?

`File handling` allows python program to read data files and write data to files permanently.

📌 Why we read files:

* Stored data permanenlty
* Read logs, reports, CSV[Comma Seperated Value]
* Process large data
* Avoid manual input

### Types of Files

1. Text Files (.txt, .csv, .log)
2. Binary files (.jpg,.pdf, .exe)

> Here we focus on reading text files.

----
## ╰┈➤ Steps to Read a File in Python

1. Open the file
2. Read the file
3. Close the File

##### ╰┈➤ Opening a File

`file = open("fileName","r")`

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
# print(fileName)
file = open(fileName,'r')  # step 1
content = file.read()  # step 2
print(content)
file.close()  # step 3
```

|Mode|Meaning|
|----|---|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`|Read & write|

### Method 1: read() --> Read Entire File


```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(fileName,'r')
content = file.read()
print(content)
file.close()
```

----
Notes:

* Reads whole file at once.
* No suitable for very large files.

### Method 2: readline() -> Read One Line



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file  = open(fileName,'r')

line = file.readline()
print(line)
print(line)
print(line)
print(line)
file.close()
```

📌 Reads only one line at a time.

#### Method 3 : readlines() --> Read all lines as list



```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file  = open(fileName,'r')

content = file.readlines()

print(content)

file.close()
```

### 6. read(n) method

## 9. Readline File Using Loop. 

Best method for large files.

Example:

```python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
file  = open(fileName,"r")

for line in file:
    print(line)

file.close()
```

## 10. Using with statement [Best Practice]

with automatically closes the file.


Syntax:

`with open("fileName","Mode") as file:`

```python
with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as file:
    print(file.read())
```

Note:

No need for file.close()

### 11. Check File Exists or not.

Using OS module

```python
import os
```

```python
if os.path.exists("C:\\Users\\User\\Desktop\\NursaryRhymes.txt"):
    print("File Exists")
else:
    print("File Not Found")
```

```python
if os.path.exists("C:\\Users\\User\\Desktop\\NursaryRhymes.txt"):
    with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as f:
        print(f.read())
    print("File Exists")
else:
    print("File Not Found")
```

### 14. File Pointer Method

tell()
`Returns current position of pointer.`

seek()
`Moves pointer to position`

### tell()

```python
file = open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r')

print(file.tell())
```

#### seek()

```python
file = open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r')

file.seek(100)
print(file.read())
```

```python
print(file.tell())
```
