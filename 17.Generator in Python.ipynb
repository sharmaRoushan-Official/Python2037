### 17. Generator in Python

#### What is a Generator?

A generator is a special function that returns an iterator and provides values one at a time using yiled instead of return.

📌 Define
> A generator generates values one by one, only when needed.


### Why Generator are needed

❌ Without Generator[list]

```python
def get_numbers():
    # print("Hello world")
    return [1,2,3,4,5]
    

get_numbers()
```

----
All values stored in memeory at once.

✅️ With Generator


```python
def get_numbers():
    for i in range(1,6):
        yield i
x = get_numbers()
x
```

```python
print(next(x))
```

```python
print(next(x))
```

### yield VS return 

|return|yield
|---|----|
|End function|Pauses function|
|Return one value| Return multiple values|
|Memory heavy|Memory efficient|

### Generator with Loop

```python
def count_up(n):
    for i in range(1,n+1):
        yield i
```

```python
for num in count_up(10):
    print(num)
```

```python
obj1 = count_up(5)
print(next(obj1))
```

### How Generator Works:

📌 Steps:

1. Function is called ➼ No execution
2. Return Generator object
3. next() resumes execution
4. Stop at yield
5. Continues from same Point

### Generator VS Normal function


----

|Feature|Generator|Function|
|----|----|----|
|used|yield|return|
|Memory|Low|High|
|Execution|Lazy|Immediate|
|Values|One at a time|All at once|

### Example: Even Numbers Generator



```python
def even_numbers(n):
    for i in range(n):
        if i%2==0:
            yield i
```

```python
obj = even_numbers(10)   # 0,2,4,6,8

print(next(obj))
```

```python
print(next(obj))
```

## Real-World Example: Reading Large File



```python
def read_file(fileName):
    with open(fileName,'r') as file:
        for line in file:
            yield line.strip()
```

### Generator with `send()` 

```python
def gen():
    value  = yield 1
    yield value

g = gen()
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(g.send(10))
```

```python
print(g.send(12))
```
