### 13. Writing Files in Python

### Write is file Writing?

__Writing Files__ means storing data permanently in a file from a Python program.

📌Used for:
* Saving user data
* Logs
* Reports
* Exproting result
* Backup files

### File Writing Modes

|Mode|Mening|
|----|----|
|`w`|write(Created new/Overwrite)|
|`a`|Append(adds data at end)|
|`w+`|Write+Read|
|`a+`|Append and Read|
|`x`| Create if already exists it show error|

### Step-by-Step Process 

1. Open the file
2. Write data in the file
3. Close the file

### 2. Writing using writelines() (Multiple Lines)

```python
lines = ["Python\n","Java\n","C++\n","Kotlin\n"]

file = open("C:\\Users\\User\\Desktop\\vedika.txt",'w')
file.writelines(lines)
file.close()
```

### 3. Best Practice: Writing using `with`



```python
with open("C:\\Users\\User\\Desktop\\vedika.txt",'w') as file:
    file.write("Using with Statement\n")
    file.write("File auto closes")
```

### 4. Writing User Input to File



```python
name = input("Enter your name: ")
age = input("Enter your age: ")

with open("C:\\Users\\User\\Desktop\\vedika.txt","w") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
with open("C:\\Users\\User\\Desktop\\vedika.txt","w") as file:
    name = input("Enter your name: ")
    age = input("Enter your age: ")
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
with open("C:\\Users\\User\\Desktop\\vedika.txt","w") as file:
    name = input("Enter your name: ")
    age = int(input("Enter your age: "))
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
numbers = [10,20,30,40,50]

with open("numbers.txt","w") as file:
    for num in numbers:
        # print(num)
        file.write(str(num)+"\n")
```

### 5. Write+Read(w+ Mode)


```python
with open("sample.txt",'w+') as file:
    file.write("Hello world")
    file.seek(0)
    print(file.read())
```

### 6. Writing using Loop (Log File )

```python
with open("log.txt",'a') as file:
    for i in range(3): #0 ,1,2
        file.write(f"Log entry {i}\n")
```

## 7. Excetption handling While Writing 

```python
try:
    with open("data.txt",'x') as file:
        file.write("Hello world")
except Exception as e:
    print(e)
```

### PART-2: CSV File Writing in Python

##### What is CSV?
* Comma Separated values
* Used in Excel, databse, reports 

##### Required Library

`import CSV`

```python
import csv
```

```python
with open("student.csv","w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["ID","Name","Marks"])
    writer.writerow([1,"Ram",99])
    writer.writerow([2,"Ramesh",85])
    writer.writerow([3,"Geeta",78])
```

Notes:

* ✴︎newline= "" avoids blank lines
* ✴︎CSV used in data Scinece and backed.
* ✴︎Very important for real projects.

```python
from openpyxl import Workbook
```

```python
wb = Workbook()

sheet = wb.active

## Create column in excel file
sheet["A1"] = "ID"
sheet["B1"] = "Name"
sheet["C1"] = "Salary"

### Add data in column in excel file
sheet.append([1,"Ram",90000])
sheet.append([2,"Ramesh",45000])
sheet.append([3,"Shyam",38000])

wb.save("employee.xlsx")
```
