### Working with JSON in Python

Python provides built-in module:

`import json`

```python
import json
```

### 𝟏. Convert Python -> JSON

```python
data = {
    "name":"Poonam",
    "age": 85,
    "course":"Programming"
}
print(data,type(data))

data['name']
```

#### Convert dictionary to JSON String

```python
json_data = json.dumps(data)  # dumps -- convert python dictionary ot json string
print(json_data)
print(type(json_data))
```

```python
json_data[0]
```

```python
json_data[1]
```

```python
json_data[2]
```

### 𝟐. Convert JSON -> Python Object 

Json.loads()

```python
json_data = '{"name": "Poonam", "age": 85, "course": "Programming"}'
print(json_data,type(json_data))


python_dict = json.loads(json_data)
print(python_dict)
print(type(python_dict))
```

```python
python_dict['name'] = "Gauri Gonte"
print(python_dict)
```

```python
# dumps() 

jsonData = json.dumps(python_dict)
print(jsonData)
```

### 𝟑. Working with JSON Files


```python
data = {
    "name":"Trupti Gaikwad",
    "age":"58"
}
print(data,type(data))
```

```python
with open("data.json","w") as file:
    json.dump(data, file,indent=4)
```

### 𝟒. Rading JSON from File

```python
with open("data.json","r") as file:
    data = json.load(file)
    print(data)
```

### dump() vs dumps()

----

|Function|Use|
|----|----|
|dump()|Write to file|
|dumps()|Convert to string|
|load()|Read from file|
|loads()| Convert string to dictionary|

### Real-Example: API

```python
import json
```

```python
response = '{"status":"success","code":200}'
print(response)
```

```python
data = json.loads(response)
if data["status"] == "success":
    print("API is Working")
else:
    print("API is not working")
```

### Handling JSON Error

```python
import json

try:
    data = json.loads("invalid json")
except json.JSONDecodeError:
    print("Invalid JSON Format")
```
